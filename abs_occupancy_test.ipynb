```python
def collapse_charges(df):
    """
    Collapse the charge states of the peptides by summing their intensities.
    """

    # 1) Define your peptide‐identity columns (everything that SHOULD stay the same across charge states)
    id_cols = [
        'Protein.Group',
        'Protein.Ids',
        'Protein.Names',
        'Genes',
        'First.Protein.Description',
        'Proteotypic',
        'Stripped.Sequence',
        'Modified.Sequence'
    ]

    # 2) Identify your intensity columns (all the “…_Phospho” and “…_Whole” columns)
    intensity_cols = [c for c in df.columns if c.endswith('_Phospho') or c.endswith('_Whole')]

    collapsed = (
        df
        .groupby(id_cols)[intensity_cols]
        .sum()
        .reset_index()
    )

    # Replace 0 with NaN in the intensity columns
    collapsed[intensity_cols] = collapsed[intensity_cols].replace(0, np.nan)

    return collapsed

# Rename the columns of pr_matrix to match the shortname format
mapping = conditions.set_index('Run')['shortname'].to_dict()

# Create an empty dataframe with the same columns as pr_matrix
occupancy = pd.DataFrame(columns=pr_matrix.columns).rename(columns=mapping)
# Remove the quant columns and replace with the replicates of the reference and experiment
intensity_cols = [col for col in occupancy.columns if col.endswith('_Phospho') or col.endswith('_Whole')]
occupancy.drop(columns=intensity_cols, inplace=True)
# Add the reference and experiment columns
for i in range(1, replicates + 1):
    occupancy[f"{experiment}_{i}"] = np.nan
    occupancy[f"{reference}_{i}"] = np.nan

# Now need a template row to fill in the occupancy dataframe
template_row = pd.Series({col: np.nan for col in occupancy.columns})

precursors = collapse_charges(pr_matrix.rename(columns=mapping))

phospho_peptides = pr_matrix[pr_matrix['Modified.Sequence'].str.contains('UniMod:21')]['Modified.Sequence'].unique()

testindex = 2
p_pep = phospho_peptides[testindex]
# Remove (UniMod:21) from the peptide sequence
p_pep_no_mod = p_pep.replace('(UniMod:21)', '')

unmod_data = precursors[precursors['Modified.Sequence'] == p_pep_no_mod]
mod_data = precursors[precursors['Modified.Sequence'] == p_pep]

# For testing the occupancy calculation
i = 1



# for i in range(1, replicates + 1):
#     ref_occupancy = 

template_row[f"{experiment}_{i}"] = None # Placeholder for the experiment occupancy


# Collapse both dataframes by summing intensities of the charge states


# for p_pep in phospho_peptides:
#     # Remove (UniMod:21) from the peptide sequence
#     p_pep_no_mod = p_pep.replace('(UniMod:21)', '')
    
#     # Get the rows corresponding to this peptide
#     rows = pr_matrix[pr_matrix['Modified.Sequence'] == p_pep]
```
